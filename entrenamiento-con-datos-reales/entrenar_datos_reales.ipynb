# **Carga de los datos**

Para este ejemplo de entrenamiento con datos reales, utilizaré la base de datos Dogs vs. Cats de la página de kaggle. Vamos a descomprimir el zip, para ello se utilizarán librerías Python del sistema (*os*) y la librería *zipfile* para descomprimir el archivo.

```python
import os
import zipfile

local_zip = './entrenamiento-con-datos-reales/dogs-vs-cats.zip'
zip_ref = zipfile.ZipFile(local_zip, 'r')
zip_ref.extractall('./entrenamiento-con-datos-reales/')
zip_ref.close()
```
